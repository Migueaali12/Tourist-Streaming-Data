# 🌍 Producer SerpAPI Multi-País - Analytics Turismo Global

Producer avanzado que permite recolectar datos de múltiples países para análisis comparativo de turismo.


### Proveedor: SerpApi

SerpApi es una API en tiempo real para acceder a los resultados de búsqueda de Google . Soluciona los problemas de alquiler de proxies, resolución de captchas y análisis de JSON.

Documentación de la API: https://serpapi.com/search-api

Búsqueda por mes: hasta 100

Registrarse en: https://serpapi.com/users/sign_up y solicitar la **API KEY**

<i>search API endpoint allows you to scrape the results from Google search engine via our SerpApi service. Head to the playground for a live and interactive demo. You can query https://serpapi.com/search using a GET request.</i>

### Usando requests (enviando parámetros)


En una petición HTTP:

- Los <b>params</b> (parámetros de consulta) suelen ir en la URL, después del endPoint.
- Los <b>headers</b> (encabezados) son información adicional que se envía junto con la petición, pero no forman parte de la URL. A menudo, los headers se usan para enviar metadatos como el tipo de contenido, la autenticación, o información específica de la aplicación. Por ejemplo:
  

headers = {<br>
    "Authorization": "",  # Token de autenticación <br>
    "Content-Type": "application/json",       # Tipo de contenido esperado <br>
    "User-Agent": "MiApp/1.0"                # Identificador de la aplicación<br>
}

```python
# 🌍 Producer SerpAPI Multi-País con Kafka
import json
import time
import logging
from datetime import datetime
from typing import Dict, List, Optional
from kafka import KafkaProducer
from kafka.errors import KafkaError

# Configurar logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

class MultiCountrySerpAPIProducer:
    def __init__(self, kafka_servers=['localhost:9092'], api_key=None):
        self.api_key = api_key or "70754b855634ed741481ef99277328d8228191bb972a8741acd8826d7505d884"
        self.endpoint = "https://serpapi.com/search"
        
        # Configuración Kafka Producer
        self.producer = KafkaProducer(
            bootstrap_servers=kafka_servers,
            value_serializer=lambda v: json.dumps(v).encode('utf-8'),
            key_serializer=lambda k: k.encode('utf-8') if k else None,
            acks='all',
            retries=3,
            retry_backoff_ms=1000
        )
        
        # 🌍 Configuración de países para análisis
        self.countries_config = {
            'ecuador': {
                'location': 'Ecuador',
                'gl': 'ec',
                'hl': 'es',
                'currency': 'USD',
                'keywords': [
                    'turismo Ecuador 2025', 'destinos Ecuador', 'Quito turismo',
                    'Galápagos tour', 'Amazonía Ecuador', 'Cuenca turismo',
                    'playas Ecuador', 'aventura Ecuador', 'ecoturismo Ecuador'
                ]
            },
            'colombia': {
                'location': 'Colombia',
                'gl': 'co',
                'hl': 'es',
                'currency': 'COP',
                'keywords': [
                    'turismo Colombia 2025', 'destinos Colombia', 'Cartagena turismo',
                    'Bogotá atracciones', 'Medellín turismo', 'San Andrés playas',
                    'Amazonas Colombia', 'café Colombia tour', 'Tayrona parque'
                ]
            },
            'peru': {
                'location': 'Peru',
                'gl': 'pe',
                'hl': 'es',
                'currency': 'PEN',
                'keywords': [
                    'turismo Perú 2025', 'Machu Picchu', 'Cusco turismo',
                    'Lima gastronomía', 'Arequipa tour', 'Nazca líneas',
                    'Amazonas Perú', 'Inca trail', 'Paracas turismo'
                ]
            },
            'chile': {
                'location': 'Chile',
                'gl': 'cl',
                'hl': 'es',
                'currency': 'CLP',
                'keywords': [
                    'turismo Chile 2025', 'Atacama desierto', 'Patagonia Chile',
                    'Santiago turismo', 'Valparaíso tour', 'Torres del Paine',
                    'vinos Chile', 'Isla de Pascua', 'Andes Chile'
                ]
            },
            'argentina': {
                'location': 'Argentina',
                'gl': 'ar',
                'hl': 'es',
                'currency': 'ARS',
                'keywords': [
                    'turismo Argentina 2025', 'Buenos Aires', 'Patagonia Argentina',
                    'Mendoza vinos', 'Bariloche turismo', 'Iguazú cataratas',
                    'tango Argentina', 'Ushuaia tour', 'Salta turismo'
                ]
            },
            'mexico': {
                'location': 'Mexico',
                'gl': 'mx',
                'hl': 'es',
                'currency': 'MXN',
                'keywords': [
                    'turismo México 2025', 'Cancún playas', 'Ciudad de México',
                    'Riviera Maya', 'Tulum ruinas', 'Oaxaca cultura',
                    'Guadalajara turismo', 'Yucatán cenotes', 'Puerto Vallarta'
                ]
            }
        }
        
        logger.info(f"✅ MultiCountrySerpAPIProducer inicializado para {len(self.countries_config)} países")
    
    def search_by_country(self, country_code: str, keyword: str, num_results: int = 10) -> Optional[Dict]:
        """Realizar búsqueda específica por país"""
        if country_code not in self.countries_config:
            logger.error(f"País {country_code} no configurado")
            return None
        
        country_config = self.countries_config[country_code]
        
        params = {
            "engine": "google",
            "format": "json",
            "q": keyword,
            "location": country_config['location'],
            "gl": country_config['gl'],
            "hl": country_config['hl'],
            "api_key": self.api_key,
            "num": num_results
        }
        
        try:
            response = requests.get(self.endpoint, params=params)
            
            if response.status_code == 200:
                data = response.json()
                
                # Enriquecer con información del país
                data['country_info'] = {
                    'country_code': country_code,
                    'country_name': country_config['location'],
                    'currency': country_config['currency'],
                    'language': country_config['hl'],
                    'search_timestamp': datetime.now().isoformat()
                }
                
                return data
            else:
                logger.error(f"Error API para {country_code}: {response.status_code}")
                return None
                
        except Exception as e:
            logger.error(f"Error en búsqueda {country_code}: {e}")
            return None
    
    def process_search_results(self, search_data: Dict, country_code: str) -> List[Dict]:
        """Procesar y estructurar resultados de búsqueda"""
        if not search_data or 'organic_results' not in search_data:
            return []
        
        processed_messages = []
        country_info = search_data.get('country_info', {})
        search_metadata = search_data.get('search_metadata', {})
        search_params = search_data.get('search_parameters', {})
        
        # Procesar resultados orgánicos
        for i, result in enumerate(search_data.get('organic_results', [])):
            message = {
                'id': f"{country_code}_{search_metadata.get('id', '')}_{i}",
                'type': 'organic_result',
                'country_code': country_code,
                'country_info': country_info,
                'search_metadata': {
                    'search_keyword': search_params.get('q', ''),
                    'search_timestamp': country_info.get('search_timestamp'),
                    'total_results': search_data.get('search_information', {}).get('total_results', 0),
                    'location': search_params.get('location', ''),
                    'language': search_params.get('hl', ''),
                    'api_id': search_metadata.get('id', '')
                },
                'title': result.get('title', ''),
                'link': result.get('link', ''),
                'snippet': result.get('snippet', ''),
                'position': result.get('position', i + 1),
                'source': result.get('displayed_link', ''),
                'thumbnail': result.get('thumbnail', ''),
                'highlighted_words': result.get('highlighted_words', [])
            }
            processed_messages.append(message)
        
        # Procesar búsquedas relacionadas
        for i, related in enumerate(search_data.get('related_searches', [])):
            message = {
                'id': f"{country_code}_related_{search_metadata.get('id', '')}_{i}",
                'type': 'related_search',
                'country_code': country_code,
                'country_info': country_info,
                'search_metadata': {
                    'search_keyword': search_params.get('q', ''),
                    'search_timestamp': country_info.get('search_timestamp'),
                    'location': search_params.get('location', ''),
                    'language': search_params.get('hl', '')
                },
                'query': related.get('query', ''),
                'link': related.get('link', '')
            }
            processed_messages.append(message)
        
        return processed_messages
    
    def send_to_kafka(self, messages: List[Dict], topic: str = 'tourism_analytics'):
        """Enviar mensajes a Kafka"""
        sent_count = 0
        
        for message in messages:
            try:
                # Usar country_code como key para distribución por particiones
                future = self.producer.send(
                    topic, 
                    key=message['country_code'],
                    value=message
                )
                
                # Verificar envío
                record_metadata = future.get(timeout=10)
                sent_count += 1
                
            except KafkaError as e:
                logger.error(f"Error enviando mensaje a Kafka: {e}")
        
        logger.info(f"📤 {sent_count}/{len(messages)} mensajes enviados a topic '{topic}'")
        return sent_count
    
    def run_multi_country_analysis(self, 
                                 countries: Optional[List[str]] = None, 
                                 keywords_per_country: int = 3,
                                 delay_between_requests: float = 2.0):
        """Ejecutar análisis para múltiples países"""
        
        if countries is None:
            countries = list(self.countries_config.keys())
        
        logger.info(f"🌍 Iniciando análisis multi-país para: {countries}")
        total_messages = 0
        
        for country_code in countries:
            if country_code not in self.countries_config:
                logger.warning(f"País {country_code} no configurado, saltando...")
                continue
            
            country_config = self.countries_config[country_code]
            country_keywords = country_config['keywords'][:keywords_per_country]
            
            logger.info(f"🚀 Procesando {country_config['location']} con {len(country_keywords)} keywords")
            
            for keyword in country_keywords:
                try:
                    # Realizar búsqueda
                    search_data = self.search_by_country(country_code, keyword)
                    
                    if search_data:
                        # Procesar resultados
                        messages = self.process_search_results(search_data, country_code)
                        
                        if messages:
                            # Enviar a Kafka
                            sent = self.send_to_kafka(messages)
                            total_messages += sent
                            
                            logger.info(f"✅ {country_code.upper()}: '{keyword}' → {sent} mensajes")
                        
                    # Delay entre requests para evitar rate limiting
                    time.sleep(delay_between_requests)
                    
                except Exception as e:
                    logger.error(f"Error procesando {country_code} - {keyword}: {e}")
                    continue
        
        logger.info(f"🎉 Análisis multi-país completado: {total_messages} mensajes totales")
        return total_messages
    
    def close(self):
        """Cerrar conexiones"""
        if self.producer:
            self.producer.close()
            logger.info("🔒 Producer Kafka cerrado")

print("✅ MultiCountrySerpAPIProducer implementado correctamente")
```

```python
# 🚀 Configuración y Ejecución del Producer Multi-País

# Inicializar el producer
producer = MultiCountrySerpAPIProducer(
    kafka_servers=['localhost:9092'],
    api_key="70754b855634ed741481ef99277328d8228191bb972a8741acd8826d7505d884"  # Tu API key
)

print("🌍 Producer Multi-País inicializado")
print(f"🎯 Países configurados: {list(producer.countries_config.keys())}")

# Mostrar configuración de países
print("\n📋 CONFIGURACIÓN DE PAÍSES:")
print("=" * 50)
for country_code, config in producer.countries_config.items():
    print(f"🏴 {config['location'].upper()} ({country_code})")
    print(f"   💰 Moneda: {config['currency']}")
    print(f"   🗣️ Idioma: {config['hl']}")
    print(f"   🔍 Keywords: {len(config['keywords'])}")
    print()
```

```python
# 🎯 Ejecutar Análisis Multi-País

print("🌎 INICIANDO ANÁLISIS MULTI-PAÍS DE TURISMO")
print("=" * 60)

# Opción 1: Analizar todos los países (recomendado para primera ejecución)
try:
    print("🚀 Ejecutando análisis completo para todos los países...")
    
    total_messages = producer.run_multi_country_analysis(
        countries=None,  # Todos los países configurados
        keywords_per_country=2,  # 2 keywords por país para empezar
        delay_between_requests=3.0  # 3 segundos entre requests
    )
    
    print(f"\n🎉 ¡ANÁLISIS COMPLETADO!")
    print(f"📊 Total de mensajes generados: {total_messages}")
    print(f"🎯 Datos enviados al topic 'tourism_analytics'")
    
except KeyboardInterrupt:
    print("\n⚠️ Análisis interrumpido por el usuario")
except Exception as e:
    print(f"\n❌ Error durante el análisis: {e}")
finally:
    producer.close()
    print("\n🔒 Producer cerrado correctamente")
```
